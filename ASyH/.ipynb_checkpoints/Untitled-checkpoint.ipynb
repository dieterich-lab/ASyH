```python
import pandas as pd
import json
```

```python
from ASyH.report import Report
```

```python
synth_data_path = "../examples/exp_ucc_250122/generis_ucc_holdout-synth.csv"

real_data_path = "../examples/exp_ucc_250122/src_holdout_ucc.csv"

metadata_path = "../examples/exp_ucc_250122/metadata_ucc.json"
```

```python
with open(metadata_path, 'r') as md_file:
    metadata = json.load(md_file)
```

```python
synth_data = pd.read_csv(synth_data_path)

real_data = pd.read_csv(real_data_path)
```

```python
synth_data_2 = synth_data.drop(columns=['Unnamed: 0.1'], axis=1)
```

```python
report = Report(real_data, synth_data_2, metadata)
```

```python
report.generate?
```

```python
import ASyH
```

```python
app = ASyH.Application()
```

```python
# Usage: Report::generate(dataset_path, sd_model_type)
# sd_model_type - class name of the best model used (found in the synthesize run of the app)
# dataset_path_root - the path to original CSV file without extension suffix

report.generate("./examples/exp_ucc_250122/src_holdout_ucc", ASyH.models.GaussianCopulaModel)
```
