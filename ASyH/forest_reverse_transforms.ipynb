```python
import pandas as pd
import numpy as np
```

```python
path_forest_arr = 'examples/forest_arr.npy'
```

```python
with open(path_forest_arr, 'rb') as np_fl:
    forest_arr = np.load(np_fl)
```

```python
def find_closest(value, discrete_vals):
    idx = np.argmin(np.abs(discrete_vals - value))
    return discrete_vals[idx]
```

```python
def make_labels_to_nums(df_col):
    unique_labels = df_col.sort_values().unique()
    label2num = {i:label for i,label in enumerate(unique_labels)}
    return label2num
```

```python
df = pd.read_excel('examples/Kaggle_Sirio_Libanes-16features.xlsx')
```

```python
col_maps = {}
for col in df.columns:
    col_maps[col] = make_labels_to_nums(df[col])
```

```python
df_forest = pd.DataFrame(forest_arr, columns=df.columns)
```

```python
discrete_values = np.array(list(col_maps['AGE_PERCENTIL'].keys()))
# normalize the values in the column AGE_PERCENTIL, by dividing by max value, then multiply by max of discrete values
df_forest['AGE_PERCENTIL'] = df_forest['AGE_PERCENTIL'].apply(lambda val: (val/max(df_forest['AGE_PERCENTIL']) * max(discrete_values)))
# df_forest['AGE_PERCENTIL'] = df_forest['AGE_PERCENTIL'].apply(lambda val: find_closest(val, discrete_values))
```

```python
def discretize_column(df, col_name, col_maps):
    discrete_values = np.array(list(col_maps[col_name].keys()))
    df[col_name] = df[col_name].apply(lambda val: (val/max(df[col_name]) * max(discrete_values)))
    df[col_name] = df[col_name].apply(lambda val: find_closest(val, discrete_values))
    df[col_name] = df[col_name].apply(lambda val: col_maps[col_name][val])
    return df
```

```python
columns_of_interest = ['AGE_ABOVE65',
                       'PATIENT_VISIT_IDENTIFIER',
                       'GENDER',
                       'ICU',
                       'AGE_PERCENTIL',
                       'DISEASE GROUPING 1',
                        'DISEASE GROUPING 2',
                        'DISEASE GROUPING 3',
                        'DISEASE GROUPING 4',
                        'DISEASE GROUPING 5',
                        'DISEASE GROUPING 6',
                        'HTN',
                        'IMMUNOCOMPROMISED',
                        'OTHER']
```

```python
df_forest = pd.DataFrame(forest_arr, columns=df.columns)
```

```python
# apply the discretize_column function to all columns of interest in the dataframe df_forest
for col in columns_of_interest:
    df_forest = discretize_column(df_forest, col, col_maps)
```
